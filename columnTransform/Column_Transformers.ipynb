```python
import pandas as pd
import numpy as np
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.read_csv(r"C:\Users\SHILA\OneDrive\Desktop\PYTHON\ML_practiceDays\Datasets\covid_toy.csv")
df.sample(5)
```

```python
df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest=train_test_split(df.drop(columns=['has_covid']),df['has_covid'],test_size=0.2)
```

```python
# Without ColumnTransformers
# adding simple imputer, to fill null values
si=SimpleImputer()
xtrain_fever=si.fit_transform(xtrain[['fever']]) # by default fills with mean value, pass 2D

xtest_fever=si.fit_transform(xtest[['fever']])
xtrain_fever.shape
```

```python
# Without ColumnTransformers
# OrdinalEncoding
oe=OrdinalEncoder(categories=[['Mild','Strong']])
xtrain_cough=oe.fit_transform(xtrain[['cough']]) # fit_transform accepts 2D values

xtest_cough=oe.fit_transform(xtest[['cough']])
xtrain_cough.shape
```

```python
# Without ColumnTransformers
# OneHotEncoder
ohe=OneHotEncoder(drop='first',sparse_output=False)
xtrain_gender_city=ohe.fit_transform(xtrain[['gender','city']])

xtest_gender_city=ohe.fit_transform(xtest[['gender','city']])
xtrain_gender_city.shape
```

```python
# extracting age
xtrain_age=xtrain.drop(columns=['gender','fever','cough','city']).values

xtest_age=xtest.drop(columns=['gender','fever','cough','city']).values
xtrain_age.shape
```

```python
# concatenate
xtrain_transformed=np.concatenate((xtrain_age,xtrain_fever,xtrain_gender_city,xtrain_cough),axis=1)

xtest_transformed=np.concatenate((xtest_age,xtest_fever,xtest_gender_city,xtest_cough),axis=1)
xtrain_transformed.shape
```

```python
# with ColumnTransformers
from sklearn.compose import ColumnTransformer
transformer=ColumnTransformer(transformers=[
    ('tf1',SimpleImputer(),['fever']),
    ('tf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('tf3',OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])
],remainder='passthrough')
```

```python
transformer.fit_transform(xtrain).shape
```
