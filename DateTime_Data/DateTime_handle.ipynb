```python
# convert to date_time
dates['date']=pd.to_datetime(dates['date'])
```

```python
dates.info() # date col data type is now datetime64
```

```python
# extract year
dates['year']=dates['date'].dt.year
```

```python
dates.head()
```

```python
# extract month
dates['month']=dates['date'].dt.month
```

```python
dates.head()
```

```python
# extract month name
dates['month_names']=dates['date'].dt.month_name()
```

```python
dates.head()
```

```python
# extract day
dates['day']=dates['date'].dt.day
```

```python
dates.head()
```

```python
# extract day of the week
dates["day_ofWeek"]=dates['date'].dt.day_of_week
```

```python
dates.head()
```

```python
# extract day name
dates['day_name']=dates['date'].dt.day_name()
```

```python
dates.drop(columns=['product_id','city_id','orders']).head() # monday= 0, tuesday= 1,.....sunday= 6
```

```python
# check if it is weekend
dates['is_weekend']=np.where(dates['day_name'].isin(['Saturday','Sunday']),1,0) # 1 if true else 0 (false)
dates.drop(columns=['product_id','city_id','orders']).head()
```

```python
# extract week
dates['weeks_no']=dates['date'].dt.isocalendar().week
dates.drop(columns=['product_id','city_id','orders']).head()
```

```python
# quarter of the year
dates['quarter']=dates['date'].dt.quarter
dates.drop(columns=['product_id','city_id','orders']).head()
```

```python
# semester of the year
dates['semester']=np.where(dates['quarter'].isin([1,2]), 1, 2)
dates.drop(columns=['product_id','city_id','orders']).head()
```

```python
# check current date time
import datetime
today=datetime.datetime.today()
today # (year, month, day, hour, minutes, seconds, miliseconds)
```

```python
today - dates['date'] # gap between today and every day of the dataset in days, with current time
```

```python
(today - dates['date']).dt.days # gap between today and every day of the dataset in days
```

```python
# months passed between current day and dataset dates
np.round((today - dates['date']) / np.timedelta64(30, 'D'),0)
```
