```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url= "https://github.com/NotAyushXD/Titanic-dataset/raw/refs/heads/master/train.csv"
df=pd.read_csv(url)
df.sample(2)
```

```python
# countplot
sns.countplot(df['Survived'])
```

```python
# piechart
df['Survived'].value_counts().plot(kind='pie',autopct='%.2f')
```

```python
# histogram
plt.hist(df['Age'],bins=10)
```

```python
# distplot
sns.displot(df['Age'])
```

```python
# boxplot
sns.boxplot(df['Fare'])
```
