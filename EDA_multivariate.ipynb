```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips=sns.load_dataset('tips')
tips.sample(5)
```

```python
# scatterplot
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'])
```

```python
# barplot
sns.barplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'])
```

```python
# boxplot
sns.boxenplot(x=tips['tip'],y=tips['sex'])
```

```python
# distplot
sns.displot(tips[tips['sex']=='Male'] ['tip'])
sns.displot(tips[tips['sex']=='Female'] ['tip'])
```

```python
# heatmap
sns.heatmap(pd.crosstab(tips['sex'],tips['smoker']))
```

```python
# clustermap
sns.clustermap(pd.crosstab(tips['sex'],tips['smoker']))
```

```python
# pairplot
sns.pairplot(tips,hue='sex')
```

```python
# flights dataset
flight=sns.load_dataset('flights')
flight.sample(4)
```

```python
# lineplot
sns.lineplot(x=flight['year'],y=flight['passengers'])
```

```python
# pivotplot & heatmap
sns.heatmap(flight.pivot_table(values='passengers',index='month',columns='year'))
```

```python
# pivottable & clustermap
sns.clustermap(flight.pivot_table(values='passengers',index='month',columns='year'))
```
