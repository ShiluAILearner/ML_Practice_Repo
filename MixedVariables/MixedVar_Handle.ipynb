```python
# extract numeric part and category parts of number col
# extract numeric part
df['number_numeric']=pd.to_numeric(df['number'],errors='coerce',downcast='integer') #errors='coerce' means: if a value cannot be converted (like "abc"), it becomes NaN. downcast='integer' tries to store the result in the smallest possible integer type (like int8, int16) instead of default int64, to save memory.
# extract categorical parts
df['number_category']=np.where(df['number_numeric'].isnull(),df['number'],np.nan) #np.where(condition, A, B), If numeric conversion failed → keep the original string from df['number'], If numeric conversion succeeded → set it to NaN.
```

```python
# separately extract and store cabin string and number
df['Cabin_num']=df['Cabin'].str.extract('(\d+)') # extracts one or more digits,  '(\d+)' is called a type of regex, it is for digit parts
df['Cabin_cat']=df['Cabin'].str[0] # extracts only first value (str)
```

```python
# extract the last bit of ticket as number
df['Ticket_num']=df['Ticket'].apply(lambda s : s.split()[-1])
df['Ticket_num']=pd.to_numeric(df['Ticket_num'],errors='coerce',downcast='integer')

# extract the first part of ticket as category
df['Ticket_cat']=df['Ticket'].apply(lambda s : s.split()[0])
df['Ticket_cat']=np.where(df['Ticket_cat'].str.isdigit(),np.nan,df['Ticket_cat'])

df.head()
```
