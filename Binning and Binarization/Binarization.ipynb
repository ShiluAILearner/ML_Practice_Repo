```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
```

```python
df=pd.read_csv(r"C:\Users\SHILA\OneDrive\Desktop\PYTHON\ML_practiceDays\Datasets\titanic.csv")[['Age','Fare','SibSp','Parch','Survived']]
df.shape
```

```python
df.dropna(inplace=True)
df.shape
```

```python
df.sample(4)
```

```python
df['Family']=df['SibSp']+df['Parch']
df.sample(1)
```

```python
df.drop(columns=['SibSp','Parch'],inplace=True)
df.sample(1)
```

```python
x=df.drop(columns=['Survived'])
y=df['Survived']
```

```python
xtrain, xtest, ytrain, ytest=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
xtrain.sample(4)
```

```python
clf=DecisionTreeClassifier()
clf.fit(xtrain,ytrain)
```

```python
ypred=clf.predict(xtest)
```

```python
accuracy_score(ytest,ypred)
```

```python
from sklearn.model_selection import cross_val_score
np.mean(cross_val_score(DecisionTreeClassifier(),x,y,cv=10,scoring='accuracy'))
```

```python
from sklearn.preprocessing import Binarizer
trf=ColumnTransformer([
    ('bin', Binarizer(copy=False),['Family'])
],remainder='passthrough')
```

```python
xtrain_trf=trf.fit_transform(xtrain)
xtest_trf=trf.transform(xtest)
```

```python
pd.DataFrame(xtrain_trf, columns=['Family','Age','Fare'])
```

```python
clf=DecisionTreeClassifier()
clf.fit(xtrain_trf,ytrain)
```

```python
ypred2=clf.predict(xtest_trf)
```

```python
accuracy_score(ytest,ypred2)
```

```python
x_trf=trf.fit_transform(x)
np.mean(cross_val_score(DecisionTreeClassifier(),x_trf,y,cv=10,scoring='accuracy'))
```
