```python
import pandas as pd
import requests
from bs4 import BeautifulSoup # used for web scraping
```

```python
header={"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/139.0.0.0 Safari/537.36"}
webpage=requests.get('https://www.ambitionbox.com/list-of-companies?page=1',headers=header).text
```

```python
soup=BeautifulSoup(webpage, 'lxml')
```

```python
company=soup.find_all('h2',class_='companyCardWrapper__companyName')
```

```python
len(company)
```

```python
rating=soup.find_all('div',class_='rating_text')
len(rating)
```
