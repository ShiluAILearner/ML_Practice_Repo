```python
import sweetviz as sv
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
iris_df=sns.load_dataset("iris")
iris_df.head()
```

```python
iris_report=sv.analyze(iris_df)
iris_report
```

```python
iris_report.show_html("iris_data_visualisation.html")
```

```python
df1=iris_df[0:70]
df2=iris_df[70:]
compare=sv.compare([df1,'df_1'],[df2,'df_2'])
compare.show_html("IrisData_comparison.html")
```
